```python
from time import sleep
from datetime import datetime
import sys
```

```python
start = datetime.now()

for i in range(21):
    sys.stdout.write('\r')
    time_spent = datetime.now() - start
    time_left  = time_spent/(i+1) * (21-i)
    sys.stdout.write(f"Progress: [{'='*i+'>':21s}] {5*i}% in {time_spent}; Estimated time left:{time_left}")
    sys.stdout.flush()
    sleep(0.1)
```

```python
from tqdm import tqdm

pbar= tqdm(range(21), bar_format="{desc:<5.5}{percentage:3.0f}%|{bar:20}{r_bar}")

for i in range(21):
    pbar.update(1)
    sleep(0.1)

pbar.close()
```
