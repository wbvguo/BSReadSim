```python
from Bio import SeqIO
```

```python
my_fa = SeqIO.to_dict(SeqIO.parse("/home/wbguo/iproject/test/GRCh38_full_analysis_set_plus_decoy_hla.fa", "fasta"))

sys.getsizeof(my_fa)
```

```python
def fa_len1():
    sum=0
    for key, seq in my_fa.items():
        sum += len(seq)
    return sum

def fa_len2():
    sum = 0 
    for key in my_fa:
        sum += len(my_fa[key])
    return sum

```

```python
timeit(fa_len1, number=1000)
```

```python
timeit(fa_len2, number=1000)
```

```python
from timeit import timeit

def a():
    x = '1'
    if isinstance(x, str):
        return(x)

def b():
    x = '1'
    if type(x) is str:
        return(x)
    
    
def c():
    x = '1'
    if x:
        return(x)
```

```python
timeit(a)
```

```python
timeit(b)
```

```python
timeit(c)
```

```python
def d():
    heter_flag = '+'
    het = True if heter_flag == '+' else False
    return(het)

heter_dict = {"+": True, "-": False}
def e():
    heter_flag = '+'
    het = heter_dict[heter_flag]
    return(het)
```

```python
timeit(d)
```

```python
timeit(e)
```
