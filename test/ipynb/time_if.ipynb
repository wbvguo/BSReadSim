# if statement judge

```python
def a():
    x = '1'
    if isinstance(x, str):
        return(x)

def b():
    x = '1'
    if type(x) is str:
        return(x)
    
    
def c():
    x = '1'
    if x:
        return(x)
```

```python
%timeit a()
```

```python
%timeit b()
```

```python
%timeit c()
```

```python
x = ['1']* 1000
y = False

def f():
    if x:
        return True

def h():
    if y:
        return True
```

```python
%timeit f()
```

```python
%timeit h()
```

# if else vs dict

```python
def d():
    heter_flag = '+'
    het = True if heter_flag == '+' else False
    return(het)

heter_dict = {"+": True, "-": False}
def e():
    heter_flag = '+'
    het = heter_dict[heter_flag]
    return(het)
```

```python
%timeit d()
```

```python
%timeit e()
```
