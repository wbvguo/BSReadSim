```python
dir_to_pretrained_model = "/home/wbguo/iproject/BSReadSim/test/code/DNABERT/pretrained/5-new-12w-0/pytorch_model.bin"
config = BertConfig.from_pretrained('https://raw.githubusercontent.com/jerryji1993/DNABERT/master/src/transformers/dnabert-config/bert-config-5/config.json')
tokenizer = DNATokenizer.from_pretrained('dna5')
model = BertModel.from_pretrained(dir_to_pretrained_model, config=config)
```

```python
sequence = "AATCTA ATCTAG TCTAGC CTAGCA"
model_input = tokenizer.encode_plus(sequence, add_special_tokens=True, max_length=512)["input_ids"]
model_input = torch.tensor(model_input, dtype=torch.long)
model_input = model_input.unsqueeze(0)   # to generate a fake batch with batch size one

output = model(model_input)
```

# encode k-mer

```python
k=7
kmer_fa = f'/home/wbguo/iproject/BSReadSim/test/kmer/{k}mer.fa'

kmer_dict = {}

with open(kmer_fa, 'r') as f:
    while True:
        line1 = int(f.readline().strip()[1:])
        line2 = f.readline().strip()
        
        # embed line2
        model_input = tokenizer.encode_plus(line2, add_special_tokens=True, max_length=512)["input_ids"]
        model_input = torch.tensor(model_input, dtype=torch.long)
        model_input = model_input.unsqueeze(0)   # to generate a fake batch with batch size one
        kmer_vector = np.squeeze(model(model_input)[1])
        kmer_dict[line2] = [kmer_vector, line1]
```
