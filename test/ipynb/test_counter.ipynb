```python
import multiprocessing
from collections import Counter

def count_words(chunk):
    counter = Counter()
    for word in chunk:
        counter[word] += 1
    return counter
```

```python
# Split the data into chunks
data = ['apple', 'banana', 'orange', 'banana', 'apple', 'orange', 'orange', 'orange', 'apple', 'banana']
chunk_size = len(data) // 3
chunks = [data[i:i+chunk_size] for i in range(0, len(data), chunk_size)]

# Create a pool of worker processes
with multiprocessing.Pool(processes=3) as pool:
    # Assign each chunk to a separate process
    results = pool.map(count_words, chunks)

# Combine the results of all the Counter objects
final_counter = Counter()
i=0
for counter in results:
    print(i)
    final_counter.update(counter)
    i+=1

print(final_counter)
```
