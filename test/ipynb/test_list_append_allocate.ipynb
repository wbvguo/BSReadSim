```python
import numpy as np
```

```python
def doAppend( size=1000 ):
    result = []
    for i in range(size):
        message= [np.random.random(size=150) for i in range(8)]
        result.append(message)
    return result

def doAllocate( size=1000 ):
    result=size*[None]
    for i in range(size):
        message= [np.random.random(size=150) for i in range(8)]
        result[i]= message
    return result
```

```python
%timeit x = doAppend()
```

```python
%timeit y = doAllocate()
```

```python
y = doAllocate()
```
