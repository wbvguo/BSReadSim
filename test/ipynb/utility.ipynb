```python
def fa_len1():
    sum=0
    for key, seq in my_fa.items():
        sum += len(seq)
    return sum

def fa_len2():
    sum = 0 
    for key in my_fa:
        sum += len(my_fa[key])
    return sum
```

```python
timeit(fa_len1, number=1000)
```

```python
timeit(fa_len2, number=1000)
```
