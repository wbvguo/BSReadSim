```python
import pysam
import multiprocessing
```

```python
def split_bam_by_tag(bam_filename, tag_value1, tag_value2):
    # Open input and output BAM files
    with pysam.AlignmentFile(bam_filename, 'rb') as input_bam, \
            pysam.AlignmentFile(f'{tag_value1}.bam', 'wb', template=input_bam) as output_bam1, \
            pysam.AlignmentFile(f'{tag_value2}.bam', 'wb', template=input_bam) as output_bam2:
        
        # Iterate over input BAM file
        for read in input_bam:
            # Check for tag value
            if read.has_tag('YS') and read.get_tag('YS') == tag_value1:
                output_bam1.write(read)
            elif read.has_tag('YS') and read.get_tag('YS') == tag_value2:
                output_bam2.write(read)
```

```python
bam_file  = "/home/wbguo/iproject/BSReadSim/test/data/RRBS/SRR6294859.sorted.bam"
tag_value1= "W_C2T"
tag_value2= "C_G2A"
```

```python
with multiprocessing.Pool(processes=1) as pool:
    pool.starmap(split_bam_by_tag, [(bam_file, tag_value1, tag_value2)]) 
```
