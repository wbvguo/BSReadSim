```python
import pandas as pd
import numpy as np
```

```python
cgmap_file = "/home/wbguo/iproject/BSReadSim/test/data/sim/pe_d/sim.CGmap.gz"
asm_file   = "/home/wbguo/iproject/BSReadSim/test/data/sim/pe_d/sim.asm.gz"
```

```python
df = pd.read_csv(cgmap_file, compression='gzip', header=None, sep='\t', quotechar='"')
```

```python
df
```

```python
asm_df = df.loc[:,[0,1,2,3,5]]
asm_df.columns = ['chr_id', 'base', 'pos', 'context', 'meth']
asm_df[['ref_meth', 'alt_meth', 'snp_pos','ref|alt','count','p_val']] = pd.DataFrame([[np.nan]*6], index=df.index)

# will not be output
asm_df['tmp'] = np.exp(np.random.rand(df.shape[0]) - 0.5)
asm_df['ref_meth'] = asm_df['meth']*2/(1+asm_df['tmp'])
asm_df['alt_meth'] = asm_df['meth']*2*asm_df['tmp']/(1+asm_df['tmp'])
asm_df['snp_pos']  = 1000             #"snp_pos"
asm_df['ref|alt']  = 'A|G'         #"ref|alt"
asm_df['count']    = '10,10,10,10' #"ref_met,ref_tot,alt_met,alt_tot"
asm_df['p_val']    = 0.05
```

```python
keep_col = ['chr_id', 'base', 'pos', 'meth', 'ref_meth', 'alt_meth', 'snp_pos','ref|alt','count','p_val']
asm_df2  = asm_df.loc[(asm_df['ref_meth'] <1) & (asm_df['alt_meth'] < 1), keep_col]
```

```python
asm_df2
```

```python
asm_df2.to_csv(asm_file, index=False, header=None, compression="gzip", quoting=None, sep='\t')
```
