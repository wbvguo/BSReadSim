# apply async

```python
import multiprocessing

def worker(shared_dict):
    # increment the counter for each process
    shared_dict['counter'] += 1

if __name__ == '__main__':
    with multiprocessing.Manager() as manager:
        shared_dict = manager.dict()
        shared_dict['counter'] = 0

        with multiprocessing.Pool(processes=4) as pool:
            results = [pool.apply_async(worker, args=(shared_dict,)) for i in range(10)]
            for result in results:
                result.wait()

        print("Counter:", shared_dict['counter'])
```

```python
import multiprocessing
from collections import Counter

def count_words(text):
    return Counter(text.split())

if __name__ == '__main__':
    # Create a pool with 4 worker processes
    with multiprocessing.Pool(processes=4) as pool:
        # Define the input data as an iterator
        input_data = iter([
            'the quick brown fox jumps over the lazy dog',
            'an apple a day keeps the doctor away',
            'a stitch in time saves nine'
        ])

        # Use imap_unordered to apply the function to each item in the input data
        # and collect the results in a list
        results = pool.imap_unordered(count_words, input_data)

        # Merge the counter objects returned by the processes
        merged_result = Counter()
        for result in results:
            merged_result += result

        # Print the final result
        print(merged_result)
```
