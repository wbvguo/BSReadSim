```python
from Bio import SeqIO
from tqdm import tqdm
import re
```

```python
ref_fasta = "/home/wbguo/iproject/BSReadSim/test/data/GRCh38_full_analysis_set_plus_decoy_hla.fa"
```

```python
ref_dict   = SeqIO.to_dict(SeqIO.parse(ref_fasta, "fasta"))
```

```python
pos = []
```

```python
for site_match in tqdm(re.finditer("CCGG", str(ref_dict['chr1'].seq.upper()))):
    pos.append(site_match.start())
```

```python
pos_num = len(pos)
```

```python
frag= []
```

```python
for i in range(pos_num):
    for j in range(i+1, pos_num):
        if pos[j] - pos[i] > 1000:
            break
        frag.append((pos[i], pos[j]))
```
