# test class method's variable scope

```python
class s:
    def __init__(self):
        self.x = 10
        self.y = 5
        self.z = {1:10, 2:5}
        
    def func(self):
        my_list = [{1:10, 2:5}, {10:10, 5:5}]
        print(my_list)
        self.func2(my_list)
        print(my_list)
        self.func3(my_list[0])
        print(my_list)
    
    def func2(self, my_list):
        my_list[0][3] = 3.33
        print(my_list)
    
    def func3(self, my_dict):
        my_dict[3] = 1
        print(my_dict)
```

```python
t = s()
```

```python
t.func()
```
