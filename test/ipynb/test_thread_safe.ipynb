```python
import random
from threading import Lock

class ThreadSafeGenerator:
    def __init__(self, generator):
        self.generator = generator
        self.lock = Lock()

    def __iter__(self):
        return self

    def __next__(self):
        with self.lock:
            return next(self.generator)

def random_number_generator():
    while True:
        yield random.randint(1, 100)

thread_safe_generator = ThreadSafeGenerator(random_number_generator())

# Example usage with multiple threads:
import threading

def print_numbers():
    for i in range(5):
        print(threading.current_thread().name, next(thread_safe_generator))

threads = [threading.Thread(target=print_numbers, name=f'Thread-{i+1}') for i in range(3)]

for thread in threads:
    thread.start()

for thread in threads:
    thread.join()
```
