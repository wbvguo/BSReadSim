```python
import statsmodels.api as sm
import numpy as np
import matplotlib.pyplot as plt

# Generate example data
np.random.seed(0)
x = 5 * np.random.rand(40, 2)
y = np.sin(x[:, 0] + x[:, 1])

# Fit the regression model
kr = sm.nonparametric.KernelReg(y, x, 'c', bw='silverman')
z, _ = kr.fit(x)

# Plot the results
plt.scatter(x[:, 0], x[:, 1], c=y)
plt.show()
```
